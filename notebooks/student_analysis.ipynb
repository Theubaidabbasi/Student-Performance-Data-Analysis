```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("All libraries imported successfully!")
```

```python

import pandas as pd

file_path = r"C:\Users\PMLS\Desktop\student dataa analysis\data sets\student-mat.csv"

df = pd.read_csv(file_path, sep=";")

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)

display(df.head())


```

```python

# 1. First five rows
print("FIRST FIVE ROWS")
display(df.head())

# 2. Last five rows
print("\nLAST FIVE ROWS")
display(df.tail())

# 3. Dataset shape
print("\nDATASET SHAPE")
print(df.shape)

# 4. Column names
print("\nCOLUMN NAMES")
print(df.columns.tolist())

# 5. Data types and non-null values
print("\nDATASET INFORMATION")
df.info()

# 6. Statistical summary
print("\nSTATISTICAL SUMMARY")
display(df.describe())

# 7. Missing values
print("\nMISSING VALUES")
print(df.isnull().sum())

# 8. Duplicate rows
print("\nDUPLICATE ROWS")
print(df.duplicated().sum())
```

```python

# Create a cleaned copy
clean_df = df.copy()

# Remove duplicate rows, if any
clean_df = clean_df.drop_duplicates()

# Check missing values
print("Missing values before handling:")
print(clean_df.isnull().sum().sum())

# Save cleaned dataset
clean_df.to_csv(
    r"C:\Users\PMLS\Desktop\student dataa analysis\cleaned_student_data.csv",
    index=False
)

print("\nCleaned dataset saved successfully!")
print("Cleaned dataset shape:", clean_df.shape)
```

```python

# STEP 4: NUMPY ANALYSIS

# Extract final grades as a NumPy array
grades = clean_df["G3"].to_numpy()

print("First 10 final grades:")
print(grades[:10])

# Calculate statistics using NumPy
print("\nMean grade:", np.mean(grades))
print("Median grade:", np.median(grades))
print("Standard deviation:", np.std(grades))
print("Minimum grade:", np.min(grades))
print("Maximum grade:", np.max(grades))

# Filter students whose final grade is 15 or above
high_grade_students = grades[grades >= 15]

print("\nStudents with grades >= 15:", len(high_grade_students))
print("Their grades:")
print(high_grade_students)

# Count students who passed (G3 >= 10)
passed_students = grades[grades >= 10]

print("\nStudents with G3 >= 10:", len(passed_students))
```

```python

# STEP 5: PANDAS ANALYSIS

# Average final grade by gender
print("Average final grade by gender:")
display(clean_df.groupby("sex")["G3"].mean().round(2))

# Average final grade by study time
print("\nAverage grade by study time:")
display(
    clean_df.groupby("studytime")["G3"]
    .agg(["mean", "median", "count"])
    .round(2)
)

# Average grade by number of absences
print("\nAverage grade by absence level:")
absence_groups = pd.cut(
    clean_df["absences"],
    bins=[-1, 0, 5, 10, float("inf")],
    labels=["0 absences", "1-5", "6-10", "More than 10"]
)

display(clean_df.groupby(absence_groups, observed=False)["G3"].mean().round(2))

# Correlation with final grade
print("\nCorrelation with final grade (G3):")
display(
    clean_df.select_dtypes(include="number")
    .corr()["G3"]
    .sort_values(ascending=False)
    .round(3)
)
```

```python

# STEP 6: OBJECT-ORIENTED PROGRAMMING (OOP)

class StudentAnalyzer:
    def __init__(self, dataframe):
        self.df = dataframe

    def show_summary(self):
        print("Total students:", len(self.df))
        print("Total columns:", len(self.df.columns))
        print("Average final grade:", round(self.df["G3"].mean(), 2))

    def grade_by_gender(self):
        return self.df.groupby("sex")["G3"].mean().round(2)


class StudentVisualizer:
    def __init__(self, dataframe):
        self.df = dataframe

    def plot_grade_distribution(self):
        plt.figure(figsize=(8, 5))
        sns.histplot(data=self.df, x="G3", bins=10, kde=True)
        plt.title("Distribution of Final Grades")
        plt.xlabel("Final Grade (G3)")
        plt.ylabel("Number of Students")
        plt.tight_layout()
        plt.show()


# Create objects
analyzer = StudentAnalyzer(clean_df)
visualizer = StudentVisualizer(clean_df)

# Use the methods
analyzer.show_summary()

print("\nAverage final grade by gender:")
display(analyzer.grade_by_gender())

visualizer.plot_grade_distribution()
```

```python

import os

output_dir = r"C:\Users\PMLS\Desktop\student dataa analysis\visualizations"
os.makedirs(output_dir, exist_ok=True)

# Graph 1: Final grade distribution
plt.figure(figsize=(8, 5))
sns.histplot(data=clean_df, x="G3", bins=10, kde=True)
plt.title("Distribution of Final Grades")
plt.tight_layout()
plt.savefig(os.path.join(output_dir, "01_grade_distribution.png"))
plt.show()
plt.close()

# Graph 2: Grade by gender
plt.figure(figsize=(7, 5))
sns.barplot(data=clean_df, x="sex", y="G3", errorbar=None)
plt.title("Average Final Grade by Gender")
plt.tight_layout()
plt.savefig(os.path.join(output_dir, "02_grade_by_gender.png"))
plt.show()
plt.close()

# Graph 3: Study time vs grade
plt.figure(figsize=(8, 5))
sns.barplot(data=clean_df, x="studytime", y="G3", errorbar=None)
plt.title("Study Time vs Final Grade")
plt.xlabel("Study Time Category")
plt.tight_layout()
plt.savefig(os.path.join(output_dir, "03_studytime_vs_grade.png"))
plt.show()
plt.close()

# Graph 4: Absences vs grade
plt.figure(figsize=(8, 5))
sns.scatterplot(data=clean_df, x="absences", y="G3", alpha=0.7)
plt.title("Absences vs Final Grade")
plt.tight_layout()
plt.savefig(os.path.join(output_dir, "04_absences_vs_grade.png"))
plt.show()
plt.close()

# Graph 5: Correlation heatmap
plt.figure(figsize=(10, 7))
numeric_df = clean_df.select_dtypes(include="number")
sns.heatmap(numeric_df.corr(), cmap="coolwarm", center=0)
plt.title("Correlation Heatmap")
plt.tight_layout()
plt.savefig(os.path.join(output_dir, "05_correlation_heatmap.png"))
plt.show()
plt.close()

print("All five visualizations saved successfully!")
print("Folder:", output_dir)
```
